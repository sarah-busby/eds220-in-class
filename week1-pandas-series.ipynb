# Pandas series

```python
import pandas as pd
import numpy as np
```

```python
# A numpy array
arr = np.random.randn(4) # random values from std normal distribution
print(type(arr))
print(arr, "\n")

# A pandas series made from the previous array
s = pd.Series(arr)
print(type(s))
print(s)
```

```python
# A series from a numpy array 
pd.Series(np.arange(3), index=[2023, 2024, 2025])
```

```python
# A series from a list of strings with default index
pd.Series(['EDS 220', 'EDS 222', 'EDS 223', 'EDS 242'])
```

```python
# Construct dictionary
d = {'key_0':2, 'key_1':'3', 'key_2':5}

# Initialize series using a dictionary
pd.Series(d)
```

```python
pd.Series(3.0, index = ['A', 'B', 'C'])
```

```python
# Define a series
s = pd.Series([98,73,65],index=['Andrea', 'Beth', 'Carolina'])

# Divide each element in series by 10
print(s /10, '\n')

# Take the exponential of each element in series
print(np.exp(s), '\n')

# Original series is unchanged
print(s)
```

```python
s > 70
```

```python
# Series with NAs in it
s = pd.Series([1, 2, np.nan, 4, np.nan])
s
```

```python
# Check if series has NAs
s.hasnans
```

```python
s.isna()
```

```python
s = pd.Series([5, -999, 8, -999], index=['A', 'B', 'C', 'D'])
print(s)
```

```python
?pd.Series.mask
```

```python
s = s.mask(s == -999)
print(s)
```

```python
# Initialize dictionary with columns' data 
d = {'col_name_1' : pd.Series(np.arange(3)),
     'col_name_2' : pd.Series([3.1, 3.2, 3.3]),
     }

# Create data frame
df = pd.DataFrame(d)
df
```

```python
# Change index
df.index = ['a','b','c']
df
```

```python
df.columns = 'C1', 'C2'
print(df)
```

## In Summary
We went over two classes in pandas:  
* **Series**  
* **Dataframe**

#### Series
* A *Series* contains an index and a single column of values
* The index by default starts at 0 and counts up, but can be set to any string or number values

#### Dataframe
* A *dataframe* contains an index and multiple rows
* Like a series, the index can be modified
* Additionally, column names can be found using `df.columns` and can also be reset this way
